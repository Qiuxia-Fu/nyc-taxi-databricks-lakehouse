This notebook is to create the catalog, schema and volume for this project.

```python
%sql
CREATE CATALOG IF NOT EXISTS nyc_taxi
COMMENT 'Project 4: NYC Yellow Taxi Lakehouse (Medallion architecture)';
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS nyc_taxi.landing COMMENT 'Raw source files, unchanged';
CREATE SCHEMA IF NOT EXISTS nyc_taxi.bronze  COMMENT 'Raw data as Delta tables + ingestion metadata';
CREATE SCHEMA IF NOT EXISTS nyc_taxi.silver  COMMENT 'Cleaned, deduplicated, typed data';
CREATE SCHEMA IF NOT EXISTS nyc_taxi.gold    COMMENT 'Business-level aggregates';
```

```python
%sql
CREATE VOLUME IF NOT EXISTS nyc_taxi.landing.raw_files
COMMENT 'NYC TLC yellow taxi parquet files 2019-01 to 2019-03, plus zone lookup';
```
