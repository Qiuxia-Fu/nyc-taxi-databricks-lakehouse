```python
%python
LANDING_PATH = "/Volumes/nyc_taxi/landing/raw_files/yellow_tripdata/"
FILES = [
    "yellow_tripdata_2019-01.parquet",
    "yellow_tripdata_2019-02.parquet",
    "yellow_tripdata_2019-03.parquet",
]
```

```python
%python
schemas = {f: dict(spark.read.parquet(LANDING_PATH + f).dtypes) for f in FILES}

all_cols = sorted(set().union(*[s.keys() for s in schemas.values()]))
for col in all_cols:
    types = {f[-15:-8]: schemas[f].get(col, "MISSING") for f in FILES}
    if len(set(types.values())) > 1:
        print(col, types)
```

```python

from pyspark.sql import functions as F

df_raw = spark.read.parquet(LANDING_PATH)

display(
    df_raw.groupBy(F.col("_metadata.file_name").alias("source_file"))
          .count()
          .orderBy("source_file")
)
```

```python
display(
    df_raw.groupBy(F.date_format("tpep_pickup_datetime", "yyyy-MM").alias("pickup_month"))
          .count()
          .orderBy("pickup_month")
)
```
